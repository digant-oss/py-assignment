```python
#11. Count number of digits in a num
n = int(input("Enter a number: "))
n = abs(n)
count = 0

if n == 0:
    count = 1
else:
    while n > 0:
        count = count + 1
        n = n // 10

print("Number of digits =", count)
```

```python
#12. Reverse a number using while loop
n = int(input("Enter a number: "))
n = abs(n)
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

print("Reverse =", rev)
```

```python
#13. Check whether a number is palindrome
n = int(input("Enter a number: "))
original = n
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

if original >= 0 and original == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#14. Print Fibonacci series up to n terms
n = int(input("Enter number of terms: "))
a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```

```python
#19. Check whether a number is prime
n = int(input("Enter a number: "))
prime = True

if n < 2:
    prime = False
else:
    for i in range(2, n):
        if n % i == 0:
            prime = False
            break

if prime:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
#23. Find sum of odd digits in a number
n = abs(int(input("Enter a number: ")))
total = 0

while n > 0:
    digit = n % 10

    if digit % 2 != 0:
        total = total + digit

    n = n // 10

print("Sum of odd digits =", total)
```

```python
#28. Check whether a number is Armstrong
n = int(input("Enter a number: "))
original = n
total = 0
digits = len(str(n))

while n > 0:
    digit = n % 10
    total = total + digit ** digits
    n = n // 10

if original >= 0 and total == original:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
#47. Find product of digits of a number
n = abs(int(input("Enter a number: ")))
product = 1

if n == 0:
    product = 0
else:
    while n > 0:
        digit = n % 10
        product = product * digit
        n = n // 10

print("Product of digits =", product)
```

```python
#49. Display menu-driven program using while loop
while True:
    print("\n--- MENU ---")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 4:
        print("Program ended")
        break

    if choice in (1, 2, 3):
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))

        if choice == 1:
            print("Answer =", a + b)

        elif choice == 2:
            print("Answer =", a - b)

        elif choice == 3:
            print("Answer =", a * b)

    else:
        print("Invalid choice")
```
