```python

# Multiplication Table and Number Analyzer

number = int(input("Enter a positive integer: "))

if number <= 0:
    print("Invalid number. Enter a positive integer.")
else:
    print("\n--- Multiplication Table ---")

    for i in range(1, 11):
        print(number, "x", i, "=", number * i)

    start = int(input("\nEnter starting number: "))
    end = int(input("Enter ending number: "))

    if start > end:
        print("Invalid range: Start cannot be greater than end")
    else:
        even_count = 0
        odd_count = 0
        total = 0

        for i in range(start, end + 1):
            total = total + i

            if i % 2 == 0:
                even_count = even_count + 1
            else:
                odd_count = odd_count + 1

        print("\n--- Number Analysis ---")
        print("Starting Number:", start)
        print("Ending Number:", end)
        print("Even Count:", even_count)
        print("Odd Count:", odd_count)
        print("Sum of Range:", total)
```
