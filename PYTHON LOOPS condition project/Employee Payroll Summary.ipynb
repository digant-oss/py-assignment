```python

# Employee Payroll Summary

employees = int(input("Enter number of employees: "))

if employees <= 0:
    print("Invalid number of employees")
else:
    total_net_payroll = 0
    valid_employees = 0

    for i in range(1, employees + 1):
        print("\nEmployee", i)

        name = input("Enter employee name: ")
        basic_salary = float(input("Enter basic salary: ₹"))
        overtime_hours = float(input("Enter overtime hours: "))

        if basic_salary < 0 or overtime_hours < 0:
            print("Invalid salary or overtime hours")
        else:
            valid_employees = valid_employees + 1

            overtime_pay = overtime_hours * 200
            gross_salary = basic_salary + overtime_pay

            if gross_salary >= 50000:
                deduction_percent = 10
            else:
                deduction_percent = 5

            deduction = gross_salary * deduction_percent / 100
            net_salary = gross_salary - deduction

            total_net_payroll = total_net_payroll + net_salary

            print("Employee Name:", name)
            print("Overtime Pay: ₹", overtime_pay)
            print("Gross Salary: ₹", gross_salary)
            print("Deduction: ₹", deduction)
            print("Net Salary: ₹", net_salary)

    print("\n--- Payroll Summary ---")
    print("Employee Count:", valid_employees)
    print("Total Net Payroll: ₹", round(total_net_payroll, 2))
```
