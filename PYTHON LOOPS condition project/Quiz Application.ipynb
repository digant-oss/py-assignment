```python

# Quiz Application

score = 0
total_questions = 5

print("Welcome to the Python Quiz!")

print("\n1. Which language are we learning?")
answer = input("Your answer: ").strip().lower()
if answer == "python":
    score = score + 1

print("\n2. What is 5 + 3?")
answer = input("Your answer: ").strip()
if answer == "8":
    score = score + 1

print("\n3. Which keyword starts a condition?")
answer = input("Your answer: ").strip().lower()
if answer == "if":
    score = score + 1

print("\n4. Which loop repeats over a sequence?")
answer = input("Your answer: ").strip().lower()
if answer == "for":
    score = score + 1

print("\n5. What is the result of 10 > 5?")
answer = input("Your answer: ").strip().lower()
if answer == "true":
    score = score + 1

percentage = (score / total_questions) * 100

print("\n--- Quiz Result ---")
print("Score:", score, "/", total_questions)
print("Percentage:", percentage, "%")

if percentage >= 80:
    print("Performance: Excellent")
elif percentage >= 60:
    print("Performance: Good")
else:
    print("Performance: Keep Practicing")
```
