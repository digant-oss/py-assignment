```python

# Shopping Cart Total

products = int(input("Enter number of products: "))

if products <= 0:
    print("Invalid number of products")
else:
    subtotal = 0
    valid_products = True

    for i in range(1, products + 1):
        print("\nProduct", i)
        price = float(input("Enter product price: ₹"))
        quantity = int(input("Enter quantity: "))

        if price < 0 or quantity < 0:
            print("Invalid price or quantity")
            valid_products = False
        else:
            item_total = price * quantity
            subtotal = subtotal + item_total
            print("Item Subtotal: ₹", item_total)

    if valid_products:
        if subtotal >= 3000:
            discount = subtotal * 10 / 100
        else:
            discount = 0

        discounted_amount = subtotal - discount

        if discounted_amount < 1500:
            delivery = 100
        else:
            delivery = 0

        final_amount = discounted_amount + delivery

        print("\n--- Shopping Cart Bill ---")
        print("Subtotal: ₹", subtotal)
        print("Discount: ₹", discount)
        print("Delivery Charge: ₹", delivery)
        print("Final Payable Amount: ₹", final_amount)
    else:
        print("Please enter valid product details and try again")
```
