```python

# Bank Transaction History

balance = float(input("Enter opening balance: ₹"))
transactions = int(input("Enter number of transactions: "))

if balance < 0:
    print("Invalid opening balance")
elif transactions < 0:
    print("Invalid number of transactions")
else:
    deposit_count = 0
    successful_withdrawals = 0
    rejected_withdrawals = 0

    for i in range(1, transactions + 1):
        print("\nTransaction", i)
        transaction_type = input(
            "Enter type (Deposit/Withdrawal): "
        ).strip().lower()

        amount = float(input("Enter amount: ₹"))

        if amount <= 0:
            print("Invalid transaction amount")

            if transaction_type == "withdrawal":
                rejected_withdrawals = rejected_withdrawals + 1

        elif transaction_type == "deposit":
            balance = balance + amount
            deposit_count = deposit_count + 1
            print("Deposit Successful")
            print("Current Balance: ₹", balance)

        elif transaction_type == "withdrawal":
            if amount <= balance:
                balance = balance - amount
                successful_withdrawals = successful_withdrawals + 1
                print("Withdrawal Successful")
                print("Current Balance: ₹", balance)
            else:
                rejected_withdrawals = rejected_withdrawals + 1
                print("Withdrawal Rejected: Insufficient Funds")

        else:
            print("Invalid transaction type")

    print("\n--- Bank Transaction Summary ---")
    print("Final Balance: ₹", balance)
    print("Deposit Count:", deposit_count)
    print("Successful Withdrawals:", successful_withdrawals)
    print("Rejected Withdrawals:", rejected_withdrawals)

    if balance < 1000:
        print("Warning: Balance is below ₹1000")
```
