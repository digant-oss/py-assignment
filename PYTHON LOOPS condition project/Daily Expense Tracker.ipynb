```python

# Daily Expense Tracker

days = int(input("Enter number of days: "))

if days <= 0:
    print("Invalid number of days")
else:
    total = 0
    highest = 0

    for day in range(1, days + 1):
        expense = float(input("Enter expense for day " + str(day) + ": "))

        if expense < 0:
            print("Invalid expense")
        else:
            total = total + expense

            if expense > highest:
                highest = expense

    average = total / days

    print("\n--- Expense Report ---")
    print("Total Expense: ₹", total)
    print("Average Daily Expense: ₹", average)
    print("Highest Daily Expense: ₹", highest)

    if total > 5000:
        print("Budget Status: Budget Alert")
    else:
        print("Budget Status: Within Budget")




```
