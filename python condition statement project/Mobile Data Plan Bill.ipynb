```python

# Mobile Data Plan Bill

name = input("Enter customer name: ")
plan = input("Enter plan (Basic/Standard/Premium): ").strip().lower()
extra_data = float(input("Enter extra data used in GB: "))

if extra_data < 0:
    print("Invalid data usage")
elif plan not in ("basic", "standard", "premium"):
    print("Invalid plan type")
else:
    if plan == "basic":
        base_charge = 299
    elif plan == "standard":
        base_charge = 499
    else:
        base_charge = 799

    extra_data_charge = extra_data * 20

    if extra_data > 10:
        surcharge = 100
    else:
        surcharge = 0

    final_bill = base_charge + extra_data_charge + surcharge

    print("\n--- Mobile Bill ---")
    print("Customer Name:", name)
    print("Plan Type:", plan.title())
    print("Base Plan Charge: ₹", base_charge)
    print("Extra Data Charge: ₹", extra_data_charge)
    print("Surcharge: ₹", surcharge)
    print("Final Bill: ₹", final_bill)
```
