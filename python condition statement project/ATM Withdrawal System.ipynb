```python

# ATM Withdrawal System

name = input("Enter account holder name: ")
balance = float(input("Enter current balance: "))
withdrawal = float(input("Enter withdrawal amount: "))

print("\n--- ATM Transaction ---")
print("Account Holder:", name)

if balance < 0:
    print("Invalid account balance")
elif withdrawal <= 0:
    print("Transaction Rejected: Amount must be greater than 0")
elif withdrawal > 20000:
    print("Transaction Rejected: Maximum limit is ₹20000")
elif withdrawal > balance:
    print("Transaction Rejected: Insufficient balance")
else:
    remaining_balance = balance - withdrawal

    print("Transaction Status: Successful")
    print("Withdrawn Amount: ₹", withdrawal)
    print("Remaining Balance: ₹", remaining_balance)

    if remaining_balance < 1000:
        print("Warning: Low Balance Warning")
```
