```python

# Movie Ticket Booking

name = input("Enter customer name: ")
age = int(input("Enter age: "))
show_type = input("Enter show type (Morning/Evening): ").strip().lower()
day_type = input("Enter day type (Weekday/Weekend): ").strip().lower()

ticket_price = 200

if age < 0:
    print("Invalid age")
elif show_type not in ("morning", "evening"):
    print("Invalid show type")
elif day_type not in ("weekday", "weekend"):
    print("Invalid day type")
else:
    if age < 12:
        discount = ticket_price * 50 / 100
    elif age >= 60:
        discount = ticket_price * 30 / 100
    else:
        discount = 0

    if show_type == "morning":
        discount = discount + 30

    if day_type == "weekend":
        surcharge = 50
    else:
        surcharge = 0

    final_price = ticket_price - discount + surcharge

    print("\n--- Movie Ticket Details ---")
    print("Customer Name:", name)
    print("Original Ticket Price: ₹", ticket_price)
    print("Total Discount: ₹", discount)
    print("Weekend Surcharge: ₹", surcharge)
    print("Final Ticket Price: ₹", final_price)
```
