```python

# Restaurant Bill and Service Charge

name = input("Enter customer name: ")
bill = float(input("Enter food bill amount: "))
membership = input("Do you have membership? (Yes/No): ").strip().lower()

if bill < 0:
    print("Invalid bill amount")
elif membership not in ("yes", "no"):
    print("Invalid membership status")
else:
    if bill >= 5000:
        discount_percent = 15
    elif bill >= 3000:
        discount_percent = 10
    elif bill >= 1000:
        discount_percent = 5
    else:
        discount_percent = 0

    discount = bill * discount_percent / 100

    if membership == "yes":
        discount = discount + (bill * 5 / 100)

    amount_after_discount = bill - discount
    service_charge = amount_after_discount * 5 / 100
    final_bill = amount_after_discount + service_charge

    print("\n--- Restaurant Bill ---")
    print("Customer Name:", name)
    print("Original Bill: ₹", bill)
    print("Discount: ₹", round(discount, 2))
    print("Service Charge: ₹", round(service_charge, 2))
    print("Final Payable Amount: ₹", round(final_bill, 2))

    if final_bill >= 5000:
        print("Thank you for being a valued customer")
    else:
        print("Thank you for visiting!")
```
